Tranformations required:
numeric_features -> StandardScaler
categorical_features -> OneHotEncoder

````python
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

#feature classification
numeric_features = ['latitude',
                    'longitude',
                    'price',
                    'minimum_nights',
                    'number_of_reviews',
                    'days_since_last_review',
                    'calculated_host_listings_count',
                    'availability_365']

categorical_features = ['neighbourhood_group',
                        'neighbourhood',
                        'room_type']

````

````python
#Transformers and pipelines
numeric_transformer = make_pipeline(
    SimpleImputer(strategy='median'),
    StandardScaler()
)

categorical_transformer = make_pipeline(
    SimpleImputer(strategy='constant', fill_value='missing'),
    OneHotEncoder(handle_unknown='ignore', sparse_output=False)
)

preprocessor = make_column_transformer(
    (numeric_transformer, numeric_features),
    (categorical_transformer, categorical_features)
)

````

````python
preprocessor.fit(X_train)
preprocessor.named_transformers_
````

````python
ohe_columns = list(
    preprocessor.named_transformers_["pipeline-2"]
    .named_steps["onehotencoder"]
    .get_feature_names_out(categorical_features)
)
new_columns = (
    numeric_features + ohe_columns
)

X_train_enc = pd.DataFrame(
    preprocessor.transform(X_train), index=X_train.index, columns=new_columns
)
X_train_enc.head()
````

````python
# Redefined preprocessor fresh for use in pipelines
preprocessor = make_column_transformer(
    (numeric_transformer, numeric_features),
    (categorical_transformer, categorical_features)
)
````

````python
#helper function for computing mean cross-validation score for each model
def mean_std_cross_val_scores(model, X_train, y_train, **kwargs):

    scores = cross_validate(model, X_train, y_train, **kwargs)

    mean_scores = pd.DataFrame(scores).mean()
    std_scores = pd.DataFrame(scores).std()
    out_col = []

    for i in range(len(mean_scores)):
        out_col.append((f"%0.3f (+/- %0.3f)" % (mean_scores.iloc[i], std_scores.iloc[i])))

    return pd.Series(data=out_col, index=mean_scores.index)
````

````python
from sklearn.dummy import DummyRegressor
from sklearn.model_selection import cross_validate
import pandas as pd

results_dict = {}

dummy_pipe = make_pipeline(preprocessor, DummyRegressor())

results_dict["dummy"] = mean_std_cross_val_scores(dummy_pipe, X_train, y_train, cv=5, return_train_score=True)

pd.DataFrame(results_dict)

````

### 1. KNN optimization

````python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from scipy.stats import randint

knn_pipe = make_pipeline(preprocessor, KNeighborsRegressor(n_jobs=-1))

param_grid_knn = {
    "kneighborsregressor__n_neighbors": [3, 5, 10, 20, 50]
}

knn_grid = GridSearchCV(
    knn_pipe,
    param_grid=param_grid_knn,
    cv=5,
    n_jobs=-1,
    return_train_score=True
)
knn_grid.fit(X_train, y_train)

print("Best params:", knn_grid.best_params_)
print("Best CV score:", knn_grid.best_score_.round(3))
````

````python
dt_pipe = make_pipeline(preprocessor, DecisionTreeRegressor(random_state=123))

param_grid_dt = {
    "decisiontreeregressor__max_depth": [2, 3, 5, 10, 15, 20, None]
}

dt_grid = GridSearchCV(
    dt_pipe,
    param_grid=param_grid_dt,
    cv=5,
    n_jobs=-1,
    return_train_score=True
)
dt_grid.fit(X_train, y_train)

print("Best params:", dt_grid.best_params_)
print("Best CV score:", dt_grid.best_score_.round(3))
````

### 3. random_forest optimization

````python
rf_pipe = make_pipeline(preprocessor, RandomForestRegressor(random_state=123, n_jobs=-1))

param_dist_rf = {
    "randomforestregressor__n_estimators": randint(50, 200),
    "randomforestregressor__max_depth": [None, 5, 10, 20, 30],
    "randomforestregressor__max_features": ["sqrt", "log2", None]
}

rf_random = RandomizedSearchCV(
    rf_pipe,
    param_dist_rf,
    n_iter=10,
    cv=5,
    return_train_score=True,
    random_state=123,
    n_jobs=-1
)
rf_random.fit(X_train, y_train)

print("Best params:", rf_random.best_params_)
print("Best CV score:", rf_random.best_score_.round(3))
````
